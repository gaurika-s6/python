```python

import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Load the dataset
url = "https://huggingface.co/shylee1/student-performance-ml-models/resolve/main/Expanded_data_with_more_features.csv"
df = pd.read_csv(url)

# Display first 5 rows
print("First 5 rows:")
display(df.head())

# 2. Check missing values
print("\nMissing values (True means missing):")
print(df.isnull())

# 3. Count missing values in each column
print("\nMissing values in each column:")
print(df.isnull().sum())

# 4. Fill missing values using Mean
df["NrSiblings"] = df["NrSiblings"].fillna(df["NrSiblings"].mean())
print("\nMean:", df["NrSiblings"].mean())

# 5. Fill missing values using Median
df["NrSiblings"] = df["NrSiblings"].fillna(df["NrSiblings"].median())
print("Median:", df["NrSiblings"].median())

# 6. Forward Fill for EthnicGroup
print("\nBefore Forward Fill:")
print(df["EthnicGroup"])

df["EthnicGroup"] = df["EthnicGroup"].ffill()

print("\nAfter Forward Fill:")
print(df["EthnicGroup"])

# 7. Label Encoding for Gender
encoder = LabelEncoder()
df["Gender_Encoded"] = encoder.fit_transform(df["Gender"])

print("\nGender Label Encoding:")
print(df[["Gender", "Gender_Encoded"]].head())
print("Classes:", encoder.classes_)

# 8. Label Encoding for EthnicGroup
encoder = LabelEncoder()
df["EthnicGroup_Encoded"] = encoder.fit_transform(df["EthnicGroup"])

print("\nEthnicGroup Label Encoding:")
print(df[["EthnicGroup", "EthnicGroup_Encoded"]].head())
print("Classes:", encoder.classes_)

# 9. One-Hot Encoding for Gender
encoded_df = pd.get_dummies(df, columns=["Gender"])

print("\nOne-Hot Encoding (Boolean values):")
display(encoded_df.head())

# 10. One-Hot Encoding with integer values
encoded_df = pd.get_dummies(
    df,
    columns=["Gender"],
    dtype=int
)

print("\nOne-Hot Encoding (0 and 1):")
display(encoded_df.head())

# 11. Display final dataset information
print("\nDataset shape:", df.shape)
print("\nFinal columns:")
print(df.columns.tolist())
```
