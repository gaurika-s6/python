```python
import os
import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# ---------------------------------------------------------
# Part 1: Step-by-Step Numerical Example (Student Marks)
# ---------------------------------------------------------

# Initial student marks: 5 students
marks = np.array([45, 48, 75, 80, 85]).reshape(-1, 1)
student_ids = ['S1', 'S2', 'S3', 'S4', 'S5']

# Compute distance matrix and linkage tree for 1D example
linked_marks = linkage(marks, method='ward')

print("=== Part 1: Student Marks Example ===")
print("Initial Data Points (Marks):", dict(zip(student_ids, marks.flatten())))

# Fit Agglomerative Clustering with K = 2
model_marks = AgglomerativeClustering(n_clusters=2, linkage='ward')
marks_clusters = model_marks.fit_predict(marks)

marks_df = pd.DataFrame({
    'Student': student_ids,
    'Marks': marks.flatten(),
    'Assigned Cluster': marks_clusters
})
print("\nFinal Groupings for Student Marks:")
print(marks_df.to_string(index=False))
print("\n" + "="*50 + "\n")

# ---------------------------------------------------------
# Part 2: Kaggle Iris Dataset Integration
# ---------------------------------------------------------

# Download dataset directly from Kaggle
path = kagglehub.dataset_download("uciml/iris")
csv_file = os.path.join(path, "Iris.csv")
df = pd.read_csv(csv_file)

print("=== Part 2: Kaggle Iris Dataset ===")
print(f"Dataset shape: {df.shape}")
print(df.head())

# Extract Sepal Length and Sepal Width features
X = df[['SepalLengthCm', 'SepalWidthCm']].values

# Step 3: Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Compute and Plot the Dendrogram
linked_iris = linkage(X_scaled, method='ward')

plt.figure(figsize=(12, 5))
dendrogram(linked_iris, truncate_mode='lastp', p=30)  # Truncated for readability
plt.title("Dendrogram - Agglomerative Hierarchical Clustering (Iris)")
plt.xlabel("Data Points / Cluster Size")
plt.ylabel("Euclidean Distance (Ward Linkage)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# Step 5: Apply Agglomerative Clustering (K = 3)
model_iris = AgglomerativeClustering(n_clusters=3, linkage='ward')
df['Cluster'] = model_iris.fit_predict(X_scaled)

# Step 6: Visualize the Clusters
plt.figure(figsize=(8, 5))
scatter = plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=df['Cluster'],
    cmap='viridis',
    s=50,
    edgecolor='k',
    alpha=0.8
)

plt.xlabel("Standardized Sepal Length")
plt.ylabel("Standardized Sepal Width")
plt.title("Agglomerative Hierarchical Clustering (K = 3)")
plt.colorbar(scatter, label='Cluster ID')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# Step 7: Display Sample Assignments
print("\nFirst 10 Data Points and Cluster Assignments:")
print(df[['SepalLengthCm', 'SepalWidthCm', 'Species', 'Cluster']].head(10))
```
